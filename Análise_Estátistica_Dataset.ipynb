### Principais observações

- O dataset possui 1025 registros, sem valores nulos (todas as colunas têm count = 1025).
- A idade média dos pacientes é de aproximadamente 54,4 anos (desvio padrão de 9,1 anos),
  variando entre 29 e 77 anos, com mediana de 56 anos.
- A distribuição do target é bem equilibrada: 51,3% dos pacientes têm presença de doença
  cardíaca (target = 1) e 48,7% não têm (target = 0).
- A proporção de homens no dataset é de 69,6%, contra 30,4% de mulheres — ou seja, a
  amostra tem bem mais homens do que mulheres.
- O colesterol (chol) tem média de 246 mg/dl, com bastante variação entre os pacientes
  (desvio padrão de ~51,6, indo de 126 a 564).
- A frequência cardíaca máxima (thalach) tem média de 149 bpm, variando de 71 a 202.

- Comparando por grupo, pacientes com doença cardíaca (target = 1) apresentam frequência
  cardíaca máxima maior (158,6 bpm) do que pacientes sem doença (139,1 bpm), uma diferença
  de quase 20 bpm que sugere que thalach tem relação forte com o diagnóstico — quanto
  maior a frequência cardíaca máxima atingida, maior a chance de o paciente estar no
  grupo com doença. Já o colesterol médio entre os dois grupos é próximo (251,3 mg/dl
  para quem não tem a doença vs 240,9 mg/dl para quem tem), uma diferença pequena que
  indica uma relação mais fraca entre colesterol e a presença de doença cardíaca —
  inclusive, contraintuitivamente, o colesterol médio é um pouco menor no grupo com
  doença.
